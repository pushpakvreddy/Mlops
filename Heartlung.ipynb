```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# sklearn modules
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# keras modules
from keras.models import Sequential
from keras.layers import Dense

# Load dataset
df = pd.read_csv("/content/sample_data/heart.csv")

# Split features and target
X = df.drop('target', axis=1)
y = df['target']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Build Neural Network
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=13))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

# Compile model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(X_train, y_train, epochs=100, batch_size=10)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
# Evaluate model
loss, acc = model.evaluate(X_test, y_test)
print("Accuracy:", acc)

# Predict sample patient
# Predict sample patient
sample = np.array([[52,1,0,125,212,0,1,168,0,1.0,2,2,3]])

prediction = model.predict(sample)

if prediction[0][0] > 0.5:
    print("Heart Disease Detected")
else:
    print("No Heart Disease")
```

```python
# Import required libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

from keras.models import Sequential
from keras.layers import Dense

# Load dataset
df = pd.read_csv("/content/sample_data/lungcancer.csv")

# Convert categorical columns
le = LabelEncoder()

df['GENDER'] = le.fit_transform(df['GENDER'])
df['LUNG_CANCER'] = le.fit_transform(df['LUNG_CANCER'])

# Features and target
X = df.drop('LUNG_CANCER', axis=1)
y = df['LUNG_CANCER']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build Neural Network
model = Sequential()
model.add(Dense(64, activation='relu', input_dim=15))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

# Compile model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(X_train, y_train, epochs=100, batch_size=10)

# Evaluate model
loss, acc = model.evaluate(X_test, y_test)
print("Accuracy:", acc)

# Example prediction
sample = np.array([[1,65,2,1,1,2,1,2,1,2,1,2,1,2,2]])

sample = scaler.transform(sample)

prediction = model.predict(sample)

if prediction[0][0] > 0.5:
    print("Lung Cancer Detected")
else:
    print("No Lung Cancer")
```
